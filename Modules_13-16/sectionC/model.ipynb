```python
from sklearn.linear_model import LinearRegression
import pandas as pd
import joblib

# Sample training data
data = pd.DataFrame({
    "distance": [1, 2, 3, 4, 5],
    "items": [1, 2, 3, 4, 5],
    "weather": [0, 1, 0, 1, 0],  # Numeric encoding for example
    "delivery_time": [15, 20, 25, 30, 35]
})

X = data[["distance", "items", "weather"]]
y = data["delivery_time"]

model = LinearRegression()
model.fit(X, y)

joblib.dump(model, "model.pkl")
print("Model saved successfully.")
```
